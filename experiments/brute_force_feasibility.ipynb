# Feasibility of Finding True Pareto Frontier with Brute Force
Count the number of possible combinations for configurations

```python
import math

def get_number_combin(num_user, num_item, k):
    num_combin_rec_per_user = math.comb(num_item, k) # combinations without replacement
    num_rec_all_users = num_combin_rec_per_user ** num_user # permutations with replacement
    print(f"{num_rec_all_users:,}")
```

```python
# Example with Jester
num_user = 1
num_item = 100
k = 10
get_number_combin(num_user, num_item, k)
```
